```python
from google_play_scraper import reviews, Sort
import pandas as pd


def crawl_reviews(app_id, app_name, total=10000):

    BATCH = 200
    all_reviews = []
    token = None

    while len(all_reviews) < total:

        result, token = reviews(
            app_id,
            lang="vi",
            country="vn",
            sort=Sort.NEWEST,
            count=BATCH,
            continuation_token=token
        )

        if not result:
            break

        all_reviews.extend(result)

        print(f"{app_name}: {len(all_reviews)} dòng")

    all_reviews = all_reviews[:total]

    data = []

    for r in all_reviews:
        data.append({
            "review_id": r["reviewId"],
            "app": app_name,
            "user": r["userName"],
            "rating": r["score"],
            "comment": r["content"].lower(),   # chuyển về lowercase
            "date": r["at"]
        })

    df = pd.DataFrame(data)

    filename = f"{app_name.lower()}_10000.csv"
    df.to_csv(filename, index=False, encoding="utf-8-sig")

    print("Xong:", filename, "| Tổng:", len(df))

    return df


# ===== CHẠY APP =====
crawl_reviews("com.duolingo", "Duolingo")
```

```python
from google_play_scraper import reviews, Sort
import pandas as pd

def crawl_reviews(app_id, app_name, total=10000):

    BATCH = 200
    all_reviews = []
    token = None

    while len(all_reviews) < total:

        result, token = reviews(
            app_id,
            lang="vi",
            country="vn",
            sort=Sort.NEWEST,
            count=BATCH,
            continuation_token=token
        )

        if not result:
            break

        all_reviews.extend(result)

        print(f"{app_name}: {len(all_reviews)} dòng")

    all_reviews = all_reviews[:total]

    data = []

    for r in all_reviews:
        data.append({
            "review_id": r["reviewId"],
            "app": app_name,
            "user": r["userName"],
            "rating": r["score"],
            "comment": (r["content"] or "").lower(),   # Handle None content and convert to lowercase
            "date": r["at"]
        })

    df = pd.DataFrame(data)

    filename = f"{app_name.lower()}_10000.csv"
    df.to_csv(filename, index=False, encoding="utf-8-sig")

    print("Xong:", filename, "| Tổng:", len(df))

    return df


# ===== CHẠY APP =====
crawl_reviews("com.busuu.android.enc", "Busuu")
```

```python
import pandas as pd
import glob

# lấy tất cả file csv trong folder
files = glob.glob("*.csv")

# đọc và gộp
df = pd.concat([pd.read_csv(f) for f in files], ignore_index=True)

# kiểm tra số dòng
print("Total rows:", len(df))

# lưu file tổng
df.to_csv("english_learning_apps_reviews.csv", index=False, encoding="utf-8-sig")

print("Đã xuất file: english_learning_apps_reviews.csv")
```

```python
# Drop rows containing Null value
df.dropna(subset=['comment'], inplace=True)
df.info()
```

```python
# Change data type of 'date' column to datetime
df['date'] = pd.to_datetime(df['date'])
df.info()
```

```python
num_unique_reviews = df['review_id'].nunique()
total_reviews = len(df)

print(f"Number of unique review_id values: {num_unique_reviews}")
print(f"Total number of reviews: {total_reviews}")

if num_unique_reviews == total_reviews:
    print("All review_id values are distinct.")
else:
    print("There are duplicate review_id values.")
```

```python
import pandas as pd

# Đếm số từ trong mỗi comment
df['word_count'] = df['comment'].str.split().apply(len)

# Giữ lại comment có >= 10 từ
df_filtered = df[df['word_count'] >= 25]

print("Original comments:", len(df))
print("Filtered comments:", len(df_filtered))
```

```python
import numpy as np

conditions = [
    (df_filtered['rating'] <= 2),
    (df_filtered['rating'] >= 4),
    (df_filtered['rating'] == 3)
]
choices = ['Negative', 'Positive', 'Neutral']

df_filtered['sentiment'] = np.select(conditions, choices, default='Unknown')

display(df_filtered.head())
```

```python
df_positive = df_filtered[df_filtered['sentiment'] == 'Positive']
df_negative = df_filtered[df_filtered['sentiment'] == 'Negative']
df_neutral = df_filtered[df_filtered['sentiment'] == 'Neutral']

print(f"Number of Positive reviews: {len(df_positive)}")
print(f"Number of Negative reviews: {len(df_negative)}")
print(f"Number of Neutral reviews: {len(df_neutral)}")
```

```python
num_positive_reviews = len(df_positive)
midpoint = num_positive_reviews // 2

df_positive_1 = df_positive.iloc[:midpoint].copy()
df_positive_2 = df_positive.iloc[midpoint:].copy()

print(f"Number of Positive reviews (Part 1): {len(df_positive_1)}")
print(f"Number of Positive reviews (Part 2): {len(df_positive_2)}")
```

```python
from googletrans import Translator
import time

translator = Translator()

def translate_comment(comment):
    try:
        # Add a small delay to avoid hitting rate limits
        time.sleep(0.1) # Wait for 100 milliseconds
        translated = translator.translate(comment, src='vi', dest='en')
        return translated.text
    except Exception as e:
        print(f"Error translating comment: {comment[:50]}... Error: {e}")
        return None

df_positive_1['comment_en'] = df_positive_1['comment'].astype(str).apply(translate_comment)

df_positive_1.head()
```

```python
df_positive_2['comment_en'] = df_positive_2['comment'].astype(str).apply(translate_comment)
df_positive_2.head()
```

```python
df_negative['comment_en'] = df_negative['comment'].astype(str).apply(translate_comment)
df_negative.head()
```

```python
df_neutral['comment_en'] = df_neutral['comment'].astype(str).apply(translate_comment)
df_neutral.head()
```
